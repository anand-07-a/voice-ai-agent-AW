```python
import os
import time
import warnings
import whisper

# Suppress FP16 warnings if running on CPU
warnings.filterwarnings("ignore", category=UserWarning)

# -------------------------------------------------------------
# Configuration
# -------------------------------------------------------------
# Dynamically get the current directory (works in Jupyter & standard scripts)
try:
    SCRIPT_DIR = os.path.dirname(os.path.abspath(__file__))
except NameError:
    SCRIPT_DIR = os.getcwd()

AUDIO_FILE_NAME = "utterance_1.wav"
AUDIO_FILE_PATH = os.path.join(SCRIPT_DIR, AUDIO_FILE_NAME)

def run_whisper_local():
    print("==================================================")
    print(" Running Local Whisper STT (System B)             ")
    print("==================================================")

    # 1. Verify audio file exists
    if not os.path.exists(AUDIO_FILE_PATH):
        print(f"❌ [ERROR] Could not find '{AUDIO_FILE_NAME}' in:")
        print(f"   {SCRIPT_DIR}")
        print("Please ensure your audio file is saved in this exact folder.")
        return

    try:
        import torch
        
        # 2. Automatically detect hardware (GPU vs CPU)
        device = "cuda" if torch.cuda.is_available() else "cpu"
        use_fp16 = torch.cuda.is_available()

        print(f"[*] Hardware Detected: {device.upper()}")
        print(f"[*] Math Precision: {'FP16 (GPU)' if use_fp16 else 'FP32 (CPU Fallback)'}")
        print(f"[*] Target Audio: {AUDIO_FILE_PATH}")

        # 3. Load the local model
        print("\nLoading local Whisper 'base' model...")
        start_time = time.time()
        model = whisper.load_model("base", device=device)

        # 4. Transcribe audio cleanly
        print(f"Transcribing '{AUDIO_FILE_NAME}'...")
        result = model.transcribe(
            AUDIO_FILE_PATH, 
            fp16=use_fp16,
            condition_on_previous_text=False,  # Prevents repetition loops
            temperature=0.0                    # Forces deterministic decoding
        )

        elapsed_time = time.time() - start_time

        print("\n" + "=" * 50)
        print(f"✅ Whisper Transcript:\n\"{result['text'].strip()}\"")
        print(f"⏱️ Time taken: {elapsed_time:.2f} seconds")
        print("=" * 50)

    except Exception as e:
        print(f"[ERROR] An unexpected error occurred: {e}")

if __name__ == "__main__":
    run_whisper_local()
```
